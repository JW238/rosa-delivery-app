```python
def late_rate(zone, time_block, promise):
    # Case 1: Specific zone and specific time block
    if zone != "all" and time_block != "all":
        times = delivery_times(zone, time_block, promise)
        late_orders = np.sum(times > promise)
        total_orders = len(times)

        return late_orders / total_orders * 100

    # Case 2: All zones and specific time block
    elif zone == "all" and time_block != "all":
        all_times = []

        for z in ZONES:
            times = delivery_times(z, time_block, promise)
            all_times.extend(times)

        all_times = np.array(all_times)
        late_orders = np.sum(all_times > promise)
        total_orders = len(all_times)

        return late_orders / total_orders * 100

    # Case 3: Specific zone and all time blocks
    elif zone != "all" and time_block == "all":
        all_times = []

        for t in TIME_BLOCKS:
            times = delivery_times(zone, t, promise)
            all_times.extend(times)

        all_times = np.array(all_times)
        late_orders = np.sum(all_times > promise)
        total_orders = len(all_times)

        return late_orders / total_orders * 100

    # Case 4: All zones and all time blocks
    else:
        all_times = []

        for z in ZONES:
            for t in TIME_BLOCKS:
                times = delivery_times(z, t, promise)
                all_times.extend(times)

        all_times = np.array(all_times)
        late_orders = np.sum(all_times > promise)
        total_orders = len(all_times)

        return late_orders / total_orders * 100
```

```python
# Test 1: Specific zone and specific time block
result_1 = late_rate("Far West", "Fri/Sat eve", 45)
print("Far West + Fri/Sat eve:", round(result_1, 1), "%")


# Test 2: All zones and specific time block
result_2 = late_rate("all", "Lunch", 45)
print("All zones + Lunch:", round(result_2, 1), "%")


# Test 3: Specific zone and all time blocks
result_3 = late_rate("Far West", "all", 45)
print("Far West + All time blocks:", round(result_3, 1), "%")


# Test 4: All zones and all time blocks
result_4 = late_rate("all", "all", 45)
print("All zones + All time blocks:", round(result_4, 1), "%")
```

```python
results = []

for zone in ZONES:
    for time_block in TIME_BLOCKS:
        rate = late_rate(zone, time_block, 45)
        results.append([rate, zone, time_block])

results.sort(reverse=True)

rank = 1

for result in results:
    print(rank, ".", result[1], "+", result[2], ":", round(result[0], 1), "%")
    rank = rank + 1
```

```python
def average_delivery_time(zone, time_block, promise):
    # Case 1: Specific zone and specific time block
    if zone != "all" and time_block != "all":
        times = delivery_times(zone, time_block, promise)
        return times.mean()

    # Case 2: All zones and specific time block
    elif zone == "all" and time_block != "all":
        all_times = []

        for z in ZONES:
            times = delivery_times(z, time_block, promise)
            all_times.extend(times)

        all_times = np.array(all_times)
        return all_times.mean()

    # Case 3: Specific zone and all time blocks
    elif zone != "all" and time_block == "all":
        all_times = []

        for t in TIME_BLOCKS:
            times = delivery_times(zone, t, promise)
            all_times.extend(times)

        all_times = np.array(all_times)
        return all_times.mean()

    # Case 4: All zones and all time blocks
    else:
        all_times = []

        for z in ZONES:
            for t in TIME_BLOCKS:
                times = delivery_times(z, t, promise)
                all_times.extend(times)

        all_times = np.array(all_times)
        return all_times.mean()
```

```python
# Test 1: Specific zone and specific time block
result_1 = average_delivery_time("Far West", "Fri/Sat eve", 45)
print("Far West + Fri/Sat eve:", round(result_1, 1), "minutes")


# Test 2: All zones and specific time block
result_2 = average_delivery_time("all", "Lunch", 45)
print("All zones + Lunch:", round(result_2, 1), "minutes")


# Test 3: Specific zone and all time blocks
result_3 = average_delivery_time("Far West", "all", 45)
print("Far West + All time blocks:", round(result_3, 1), "minutes")


# Test 4: All zones and all time blocks
result_4 = average_delivery_time("all", "all", 45)
print("All zones + All time blocks:", round(result_4, 1), "minutes")
```

### Part I-e: Late Rate vs. Average Delivery Time

The late rate is more relevant to Rosa's decisions than the average delivery time. Rosa currently promises customers a 45-minute delivery time, so the key question is whether orders are delivered within that promised time.

The late rate directly measures the percentage of orders that exceed the 45-minute promise. For example, the results from Part I-b show that Far West during Fri/Sat evening has the highest late rate at approximately 43.8%. This indicates that a large proportion of orders in this zone and time block are delivered later than promised.

Average delivery time is still useful for understanding overall delivery performance, but it does not directly show whether Rosa is meeting her promise. Therefore, late rate is a more relevant measure for evaluating Rosa's delivery promise and making decisions about how the promise should be changed.

```python
def late_order_cost(costs):
    return costs["refund"] + costs["churn_orders"] * costs["margin"]

cost = late_order_cost(COSTS)
print("Total cost per late order:", cost)
```

```python
def best_promise(zone, time_block, promises, costs):
    best_promise = 0
    best_profit = 0

    late_cost = costs["refund"] + costs["churn_orders"] * costs["margin"]

    for promise in promises:
        times = delivery_times(zone, time_block, promise)

        orders = len(times)
        late_orders = np.sum(times > promise)

        profit = orders * costs["margin"]
        late_costs = late_orders * late_cost

        net_profit = profit - late_costs

        print(promise, "minutes:", round(net_profit, 2))

        if net_profit > best_profit:
            best_profit = net_profit
            best_promise = promise

    print("Best promise:", best_promise, "minutes")
    print("Highest net profit:", round(best_profit, 2))

    return best_promise, best_profit

    promises = list(range(20, 71, 5))

best, profit = best_promise(
    "Far West",
    "Fri/Sat eve",
    promises,
    COSTS
)
```
